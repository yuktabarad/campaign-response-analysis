## 3. Statistical Hypothesis Testing

### Objective

The objective of this notebook is to use statistical hypothesis testing to investigate whether customer characteristics are associated with term deposit subscription rates.

### Research Questions

* Is subscription status associated with customer job category?
* Is subscription status associated with marital status?
* Does the subscription rate differ between customers with and without housing loans?

### Statistical Approach

* Formulate null and alternative hypotheses.
* Select appropriate statistical tests.
* Calculate test statistics and p-values.
* Interpret the results using a 5% significance level.

### Tools

Python, Pandas, SciPy, and Statsmodels.

### Statistical Test 1: Job Category vs Subscription Status

**Research Question:** Is a customer's job category associated with their term deposit subscription status?

**Hypotheses:**

* **Null hypothesis (H₀):** Job category and subscription status are independent. There is no statistically significant association between them.
* **Alternative hypothesis (H₁):** Job category and subscription status are associated.

**Statistical Test:** Chi-Square Test of Independence

**Significance Level:** α = 0.05

**Reason for Selecting This Test:**

Both `job` and `y` are categorical variables. The Chi-Square Test of Independence is appropriate for determining whether an association exists between two categorical variables.



```python
job_contingency_table=pd.crosstab(df['job'],df['y'])
job_contingency_table
```

```python
job_chi2, job_p_value, job_dof, job_expected = chi2_contingency(job_contingency_table)

print("Chi-square statistic:",job_chi2)
print("P-value:",job_p_value)
print("Degrees of freedom:",job_dof)
```

```python
alpha=0.05

if job_p_value < alpha:
    print("Reject H0: Job category is associated with subscription status.")
else:
    print("Fail to reject H0: There is insufficient evidencs of association.")
```

### Interpretation: Association Between Job Category and Subscription

The Chi-Square Test of Independence was conducted to investigate whether job category is associated with term deposit subscription status.

The test produced a Chi-Square statistic of 836.11 with 11 degrees of freedom and a p-value of approximately 3.34 × 10⁻¹⁷².

Since the p-value is below the significance level of 0.05, we reject the null hypothesis of independence.

**Conclusion:** There is strong statistical evidence of an association between job category and term deposit subscription status in this dataset.

This result does not establish that job category causes a customer to subscribe. Further analysis of subscription rates across job categories is needed to understand the nature of the association.
