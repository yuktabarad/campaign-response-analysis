```python
cat_cols = df.select_dtypes(include="object").columns

for col in cat_cols:
    print(f"\nColumn: {col}")
    print(df[col].unique())
```

```python
for col in cat_cols:
    unknown_count=(df[col]=='unknown').sum()

    if unknown_count>0:
        print(f"{col}: {unknown_count}")
```
